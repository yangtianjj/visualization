```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python

x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([2, 4, 3, 5, 7, 8])

fig, ax = plt.subplots(figsize=(5, 3))

# Color the figure and axis background
fig.patch.set_facecolor("lightcoral")   # fig container
ax.set_facecolor("lightgreen")          # ax plotting area

ax.scatter(x, y)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Scatter Plot")

plt.show()
```

```python
np.random.seed(613)
x = np.arange(40)
y1 = np.random.randint(0, 100,40)
y2 = np.random.randint(0, 100,40)
```

```python
fig, ax = plt.subplots()

ax.axis([0, 10, 0, 10])

# A scatter point using data coordinates
ax.scatter(1, 5)

# Text using data coordinates
ax.text(1, 5, " Data: (1, 5)", transform=ax.transData)

# Text using axes coordinates
ax.text(0.5, 0.1, " Axes: (0.5, 0.1)", transform=ax.transAxes)

# Text using figure coordinates
ax.text(0.2, 0.2, " Figure: (0.2, 0.2)", transform=fig.transFigure)

plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 4, 5, 7, 8])
y2 = np.array([1, 3, 4, 6, 7])

# Font setting
font1 = {
    'fontfamily': 'serif',
    'color': 'indigo'
}

fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

ax.legend(loc="lower right")

ax.set_xlabel("Shiny New X Axis", fontsize=18, fontdict=font1)
ax.set_ylabel("Y Axis", fontsize=14)
ax.set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```
