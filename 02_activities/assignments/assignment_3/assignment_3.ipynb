```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python


DATA_URL = "https://tor.publicbikesystem.net/ube/gbfs/v1/en/station_information"

response = requests.get(DATA_URL)
response.raise_for_status()

data = response.json()

stations = pd.DataFrame(data["data"]["stations"])

stations.to_csv("bike_share_stations_raw.csv", index=False)

print(stations.head())
print(stations.columns)
print(stations.info())

df = stations[["station_id", "name", "lat", "lon", "capacity", "is_charging_station"]].copy()

df["capacity"] = pd.to_numeric(df["capacity"], errors="coerce")
df = df.dropna(subset=["lat", "lon", "capacity"])

df.to_csv("bike_share_stations_cleaned.csv", index=False)


plt.figure(figsize=(8, 7))

plt.scatter(
    df["lon"],
    df["lat"],
    s=df["capacity"] * 2,
    alpha=0.5
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Bike Share Toronto Station Locations by Capacity")

plt.figtext(
    0.5,
    0.01,
    "Each point represents a Bike Share station. Larger points indicate higher station capacity.",
    ha="center",
    fontsize=9
)

plt.tight_layout()
plt.savefig("visualization_1_python_station_capacity_map.png", dpi=300)
plt.show()
```

```python
top_capacity = df.sort_values("capacity", ascending=False).head(15)

top_capacity[["name", "capacity"]].to_csv(
    "visualization_2_excel_top_station_capacity.csv",
    index=False
)

print(top_capacity[["name", "capacity"]])
```
